**[Task 2.2.]()** Implement the two functions `sobel_h` and `sobel_v` for horizontal and vertical $3 \times 3$ Sobel filtering, respectively:

```python
def sobel_h(img_in):
    result = numpy.zeros(img_in.shape)
    n = 1
    kernel = numpy.array([
        [-1, 0, 1],
        [-2, 0, 2],
        [-1, 0, 1]
    ])
    for p in numpy.ndindex(img_in.shape):
        i, j = p
        if i < n or j < n or i >= img_in.shape[0] - n or j >= img_in.shape[1] - n:
            continue 
        value = 0
        for q in numpy.ndindex(3, 3):
            di = q[0] - n
            dj = q[1] - n
            neighbor_i = i + di
            neighbor_j = j + dj
            value += img_in[neighbor_i, neighbor_j] * kernel [q[0], q[1]]
        
        result[p] = value/8.0
        
    return result
    
def sobel_v(img_in):
    result = numpy.zeros(img_in.shape)
    n = 1
    kernel = numpy.array([
        [-1, -2, -1],
        [0, 0, 0],
        [1, 2, 1]
    ])
    for p in numpy.ndindex(img_in.shape):
        i, j = p
        if i < n or j < n or i >= img_in.shape[0] - n or j >= img_in.shape[1] - n:
            continue 
        value = 0
        for q in numpy.ndindex(3, 3):
            di = q[0] - n
            dj = q[1] - n
            neighbor_i = i + di
            neighbor_j = j + dj
            value += img_in[neighbor_i, neighbor_j] * kernel [q[0], q[1]]
        
        result[p] = value/8.0
        
    return result
```

**[Task 2.3.]()** Test your implementations from above:
